```python
import cv2 as cv
import tensorflow as tf
from tensorflow.keras.applications.mobilenet_v2 import  MobileNetV2,preprocess_input,decode_predictions
import numpy as np

model = MobileNetV2(weights = 'imagenet')

def predict_frame(frame):
    img = cv.resize(frame,(224,224))
    img = cv.cvtColor(img,cv.COLOR_BGR2RGB)
    x = np.expand_dims(img,axis=0)
    x= preprocess_input(x)
    prediction = model.predict(x)
    
    _,result,prob = decode_predictions(prediction,top=1)[0][0]
    return result,prob

frame_count = 0
prediction_label = "Scanning..."
confidence = 0

cap = cv.VideoCapture(0)

while True:
    ret,frame = cap.read()
    if not ret:
        break
    frame_count +=1
    
    if frame_count%3 == 0:
        label, prob = predict_frame(frame)

        print(prob)
        if prob > 0.15:
            prediction_label = label.replace('_', ' ').title()
            confidence = prob * 100
        
    color = (0,255,0) if confidence > 70 else (0,165,255)
    text = f"{prediction_label}({confidence})"
    cv.putText(frame,text,(20,50),cv.FONT_HERSHEY_COMPLEX,1,color,2)
    cv.imshow("fazool",frame)
    
    key = cv.waitKey(1) & 0xFF
    if key == ord('q') or key == 27:
        break
cap.release()
cv.destroyAllWindows()
```
